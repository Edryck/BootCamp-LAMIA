```python
# Importa a biblioteca.
import spacy
```

```python
# Mesmo exemplo da aula anterior: cidade inventada e filme 'Mr. Deeds'.
nlp = spacy.load("en_core_web_sm")
text = "West Chestertenfieldville was referenced in Mr. Deeds."
```

```python
# Processa a frase.
doc = nlp(text)
```

```python
# Entidades que o modelo pré-treinado acha sozinho.
for ent in doc.ents:
    print (ent.text, ent.label_)
```

```python
# EntityRuler no fim do pipeline (depois do ner).
ruler = nlp.add_pipe("entity_ruler")
```

```python
# Confere a ordem dos componentes.
nlp.analyze_pipes()
```

```python
# Desta vez os padrões ficam numa variável separada, o que deixa tudo mais organizado.
patterns = [
        {"label": "GPE", "pattern": "West Chestertenfieldville"}
]
```

```python
# Registra os padrões no ruler.
ruler.add_patterns(patterns)
```

```python
# Reprocessa em um doc novo (doc2). Com o ruler depois do ner, o resultado pode continuar igual ao
# anterior.
doc2 = nlp(text)
for ent in doc2.ents:
    print (ent.text, ent.label_)
```

```python
# Modelo limpo (nlp2) pra refazer o teste.
nlp2 = spacy.load("en_core_web_sm")
```

```python
# Agora o ruler entra ANTES do ner.
ruler = nlp2.add_pipe("entity_ruler", before="ner")
```

```python
# Reaproveita a mesma lista de padrões.
ruler.add_patterns(patterns)
```

```python
# Processa o texto com o nlp2.
doc = nlp2(text)
```

```python
# Lista as entidades: a cidade inventada deve vir como GPE.
for ent in doc.ents:
    print (ent.text, ent.label_)
```

```python
# Confere a ordem do pipeline: entity_ruler antes do ner.
nlp2.analyze_pipes()
```

```python
# Outro modelo limpo (nlp3), pra testar com dois padrões.
nlp3 = spacy.load("en_core_web_sm")
```

```python
# Ruler antes do ner.
ruler = nlp3.add_pipe("entity_ruler", before="ner")
```

```python
# Dois padrões: a cidade (GPE) e o filme (rótulo customizado FILM).
patterns = [
        {"label": "GPE", "pattern": "West Chestertenfieldville"},
        {"label": "FILM", "pattern": "Mr. Deeds"}
]
```

```python
# Registra os padrões.
ruler.add_patterns(patterns)
```

```python
# Processa com o nlp3.
doc = nlp3(text)
```

```python
# Lista as entidades: cidade e filme.
for ent in doc.ents:
    print (ent.text, ent.label_)
```
