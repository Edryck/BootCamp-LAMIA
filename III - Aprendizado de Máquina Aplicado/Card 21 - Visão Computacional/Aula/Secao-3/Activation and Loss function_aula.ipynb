```python
import math

# (1/(1+e^(-x)))
def sigmoid(x):
    return 1 / (1 + math.exp(-x))
```

```python
import torch

class ReLU(torch.nn.Module):  # toda camada ou ativação personalizada no torch herda de nn.Module
    def __init__(self):
        super().__init__()  # sempre chama o __init__ da classe mãe primeiro

    def forward(self,x):  # forward é o método que define o que a camada faz quando chama model(x)
        return torch.max(torch.zeros_like(x), x)  # ReLU = max(0, x); zeros_like cria um tensor de zeros do mesmo shape/device que x
```

```python
import numpy as np

def mse_loss(y_true, y_pred):
    return np.mean((y_true - y_pred)**2)  # erro quadrático médio, equivale ao nn.MSELoss() do torch
```

```python
import numpy as np

def cross_entropy_loss(y_true, y_pred):
    epsilon = 1e-15
    # evita log(0) que dá -infinito
    y_pred = np.clip(y_pred, epsilon, 1. - epsilon)
    return -np.mean(y_true * np.log(y_pred) + np.log(1 - y_pred))
```
