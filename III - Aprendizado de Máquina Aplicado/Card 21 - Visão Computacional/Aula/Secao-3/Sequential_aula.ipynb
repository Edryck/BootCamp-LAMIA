```python
import torch
import torch.nn as nn
import numpy as np
from torch.utils.data import Dataset, DataLoader
device = 'cuda' if torch.cuda.is_available() else 'cpu'
```

```python
x = [[1,2],[3,4],[5,6],[7,8]]
y = [[3],[7],[11],[15]]
```

```python
class MyDataset(Dataset):
    def __init__(self,x,y):
        self.x = torch.tensor(x).float().to(device)
        self.y = torch.tensor(y).float().to(device)
    def __len__(self):
        return len(self.x)
    def __getitem__(self,ix):
        return self.x[ix], self.y[ix]
```

```python
ds = MyDataset(x,y)
dl = DataLoader(ds, batch_size=2, shuffle=True)
```

```python
model = nn.Sequential(nn.Linear(2,8), # aqui sim, um jeito bem parecido com o Sequential do keras
                      nn.ReLU(),
                      nn.Linear(8,1)).to(device)  # com Sequential também precisa mandar pro device manualmente
```

```python
# printar o modelo mostra a arquitetura resumida
model
```

```python
loss_func = nn.MSELoss()
from torch.optim import SGD
opt = SGD(model.parameters(), lr = 0.001)
import time
loss_history = []
start = time.time()

model.train() # avisa o modelo que esta em treino
for _ in range(50):
    for ix, iy in dl:
        opt.zero_grad()
        loss_value = loss_func(model(ix),iy)
        loss_value.backward()
        opt.step()
        loss_history.append(loss_value)
end = time.time()
# só cronometra quanto tempo o treino demorou
print(end - start)
```
