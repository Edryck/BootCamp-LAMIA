```python
from torchvision import datasets
import torch
from torch.utils.data import Dataset, DataLoader
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
device = 'cuda' if torch.cuda.is_available() else 'cpu'
data_folder = '/content'
fmnist = datasets.FashionMNIST(data_folder,
                               download=True,
                               train=True)
tr_images = fmnist.data
tr_targets = fmnist.targets
val_fmnist = datasets.FashionMNIST(data_folder,
                                   download=True,
                                   train=False)
val_images = val_fmnist.data
val_targets = val_fmnist.targets
```

```python
from imgaug import augmenters as iaa  # lib de data augmentation
aug = iaa.Sequential([
    iaa.Affine(translate_px={'x':(-10,10)},  # desloca a imagem até 10px pra esquerda/direita aleatoriamente
               mode='constant')
])
```

```python
class FMNISTDataset(Dataset):
  def __init__(self, x, y, aug=None):
    self.x, self.y, = x, y
    self.aug = aug
  def __getitem__(self, ix):
    x, y = self.x[ix], self.y[ix]
    return x, y
  def __len__(self): return len(self.x)

  def collate_fn(self, batch):  # collate_fn deixa você customizar como os itens viram um batch
    'logic to modify a batch of image'
    ims, classes = list(zip(*batch))
    if self.aug:
      ims_np = np.array([tensor.numpy() for tensor in ims])
      ims=self.aug.augment_images(images=ims_np)  # aplica o augmentation no batch inteiro de uma vez


    ims = torch.tensor(ims)[:,None,:,:].to(device)/255.  # [:,None,:,:] adiciona a dimensão de canal (1 canal, escala de cinza) e normaliza
    classes = torch.tensor(classes).to(device)
    return ims, classes
```

```python
from torch.optim import SGD, Adam
def get_model():
  model = nn.Sequential(
      nn.Conv2d(1, 64, kernel_size=3),
      nn.MaxPool2d(2),
      nn.ReLU(),
      nn.Conv2d(64, 128, kernel_size=3),
      nn.MaxPool2d(2),
      nn.ReLU(),
      nn.Flatten(),
      nn.Linear(3200, 256),  # 3200 é o tamanho já achatado depois das convoluções+poolings
      nn.ReLU(),
      nn.Linear(256, 10)
  ).to(device)
  loss_fn = nn.CrossEntropyLoss()
  optimizer = Adam(model.parameters(), lr=1e-3)
  return model, loss_fn, optimizer
```

```python
def train_batch(x, y, model, opt, loss_fn):
  model.train()
  prediction = model(x)
  batch_loss = loss_fn(prediction, y)
  batch_loss.backward()
  optimizer.step()
  optimizer.zero_grad()
  return batch_loss.item()
```

```python
def get_data():
  train = FMNISTDataset(tr_images, tr_targets, aug=aug)  # passa o augmentation só no dataset de treino
  'notice the collate_fn argument'
  trn_dl = DataLoader(train, batch_size=64,
                      collate_fn=train.collate_fn, shuffle=True)
  val = FMNISTDataset(val_images, val_targets)  # validação sem augmentation
  val_dl = DataLoader(val, batch_size=len(val_images),
                      collate_fn=val.collate_fn, shuffle=True)
  return trn_dl, val_dl
```
