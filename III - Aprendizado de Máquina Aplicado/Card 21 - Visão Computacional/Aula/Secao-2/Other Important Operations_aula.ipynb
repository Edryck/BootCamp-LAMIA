```python
import torch
```

```python
# shape (2,4)
x = torch.tensor([[1,2,3,4],[5,6,7,8]])
# shape (4,3)
y = torch.tensor([[1,2,3],
                  [2,3,4],
                  [4,5,6],
                  [7,8,9]])
```

```python
# multiplicacao de matriz de verdade, nao e elemento a elemento, resultado (2,3)
torch.matmul(x,y)
```

```python
# tensor aleatório shape (1,4,5)
x = torch.randn(1,4,5)
# concatena os dois no eixo 1, dobra aquela dimensao
z = torch.cat([x,x], axis=1)    
print('Concatenated axis 1:', x.shape, z.shape)
```
