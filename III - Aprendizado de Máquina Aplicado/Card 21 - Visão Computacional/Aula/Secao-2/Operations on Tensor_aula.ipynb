```python
a = torch.ones(2, 1, 10)  # tensor preenchido de 1, shape (2,1,10)
```

```python
a
```

```python
print(f"This is the shape before squeezing {a.shape}")
squeezed = a.squeeze(1)  # squeeze remove uma dimensão se ela tiver tamanho 1 aqui a dim 1 tem tamanho 1 entao some
print(f"This is the shape after squeezing {squeezed.shape}")
```

```python
a = torch.ones(2, 2, 10)
print(f"This is the shape before squeezing {a.shape}")
squeezed = a.squeeze(1)  # aqui a dim 1 tem tamanho 2, nao 1 ent o squeeze nao faz nada e o shape continua igual
print(f"This is the shape after squeezing {squeezed.shape}")
```

```python
a = torch.ones(2, 3, 1)
print(f"This is the shape before squeezing {a.shape}")
squeezed = a.squeeze(2)  # ultima dim tem tamanho 1 ent some e sobra (2,3)
print(f"This is the shape after squeezing {squeezed.shape}")
```

```python
a = torch.ones(2,3)
print(f"This is the shape before squeezing {a.shape}")
squeezed = a.unsqueeze(2)  # unsqueeze é o contrário adiciona uma dimensao de tamanho 1 na posição indicada
print(f"This is the shape after squeezing {squeezed.shape}")
```

```python
a = torch.ones(2,3)
```

```python
# indexar com None é um atalho pro unsqueeze(0), vira (1,2,3)
print(a[None].shape)
# None na posição 1 unsqueeze(1) vira (2,1,3)
print(a[:,None].shape)
# None na posição 2 unsqueeze(2) vira (2,3,1)
print(a[:,:,None].shape)
```
