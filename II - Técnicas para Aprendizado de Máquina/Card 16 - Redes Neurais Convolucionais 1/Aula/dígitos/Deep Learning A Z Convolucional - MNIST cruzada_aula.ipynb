```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
import numpy as np
from sklearn.model_selection import StratifiedKFold
```

```python
# Carrega o conjunto de dados MNIST
(X, y), (X_teste, y_teste) = mnist.load_data()
# Faz o reshape dos dados para o formato que a rede neural espera
X = X.reshape(X.shape[0], 28, 28, 1)
# Normaliza os dados para um intervalo entre 0 e 1
X = X.astype('float32')
X /= 255
# Faz a codificacao das classes para o formato one-hot encoding
y = np_utils.to_categorical(y, 10)
```

```python
np.zeros(5)
```

```python
y.shape, y.shape[0]
```

```python
# Cria um array de zeros com o mesmo numero de linhas que o array y e uma coluna
np.zeros(shape = (y.shape[0], 1))
```

```python
# Seed para sempre obter os mesmos resultados
seed = 5
np.random.seed(seed)
```

```python
# Cria o objeto StratifiedKFold com 5 splits, shuffle ativado e a seed definida
kfold = StratifiedKFold(n_splits = 5, shuffle=True, random_state = seed)
```

```python
# Loop para cada split do StratifiedKFold
resultados = []
for indice_treinamento, indice_teste in kfold.split(X, np.zeros(shape = (y.shape[0], 1))):
  print("Índices treinamento: ", indice_treinamento, "Índice teste: ", indice_teste)
  # Cria a rede neural convolucional
  classificador = Sequential()
  # Adiciona as camadas da rede neural convolucional
  classificador.add(InputLayer(shape=(28, 28, 1)))
  classificador.add(Conv2D(32, (3, 3), activation='relu'))
  classificador.add(MaxPooling2D(pool_size=(2,2)))
  classificador.add(Flatten())
  classificador.add(Dense(units=128, activation='relu'))
  classificador.add(Dense(units=10, activation='softmax'))
  # Compila a rede neural convolucional
  classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
  # Treina a rede neural convolucional com os dados de treinamento e avalia a precisão com os dados de teste
  classificador.fit(X[indice_treinamento], y[indice_treinamento], batch_size = 128, epochs = 5)
  # Avalia a precisão da rede neural convolucional com os dados de teste e armazena o resultado
  precisao = classificador.evaluate(X[indice_teste], y[indice_teste])
  resultados.append(precisao[1])
```

```python
resultados
```

```python
# Media das precisoes obtidas em cada split do StratifiedKFold
np.array(resultados).mean()
```

```python
# Desvio padrao das precisoes obtidas em cada split
np.array(resultados).std()
```
